## Lab 3. Employee data
This lab exercise makes updates to
 
information extracted from
an employee database.

```python
hr_list = [
('0123', 'HR', 'Rebecca Yang', '69000'),
('0121', 'IT', 'Mark Blick', '67000'),
('0068', 'IT', 'Kahna Larsen', '112000'),
('0234', 'OPS', 'Jim Smith', '54000')
]
```

```python
#Converting the tupul into a list
new_hr_list = list(hr_list)

#Getting Marks info and makes it a list
mark_info = list(new_hr_list[1])

#splits the names into first and last name
mark_first, mark_last = mark_info[2].split(" ")

#changes Mark's name to his new last name
mark_info[2] = mark_first + " Blick-Hawley"

#putting it back as tuple
updated_mark_info = tuple(mark_info)

#updating the tuple
new_hr_list[1] = updated_mark_info
```

```python
#Getting jim's info and makes it a list
jim_info = list(new_hr_list[3])

#Chaning Jim's Department code
jim_info[1] = "CS"

#Changing Jim's salary
jim_info[3] = "60000"

#Putting it back as tuple
updated_jim_info = tuple(jim_info)

new_hr_list[3] = updated_jim_info
                
                
```

```python
print(new_hr_list)
```

```python
for x in new_hr_list:
    employee_number = x[0]
    dept_code = x[1]
    name = x[2]
    salary = float(x[3])
    print(f"{employee_number} | {dept_code} | {name} | {salary:,}")
    
```
