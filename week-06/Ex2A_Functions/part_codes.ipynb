# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(info):
    supplier_code = info.split(":")
    supplier = supplier_code[0]
    return supplier
```

```python
def get_product_num(info):
    product_split = info.split(":")
    product_after_f_split = product_split[-1]
    product_split_s = product_after_f_split.split("-")
    product_after_s_split = product_split_s[0]
    return product_after_s_split
```

```python
def get_size(info):
    size_info = info.split("-")
    size = size_info[-1]
    if size == "L":
        converted_size = "large"
    elif size == "M":
        converted_size = "medium"
    elif size == "S":
        converted_size = "small"
    else:
        converted_size = size

    return size
```

```python
product_info = "BEARS:1997-L"
supplier = get_supplier_code(product_info)
product_num = get_product_num(product_info)
size = get_size(product_info)

print(f"supplied by {supplier}, part#{product_num},size {size}")
```
