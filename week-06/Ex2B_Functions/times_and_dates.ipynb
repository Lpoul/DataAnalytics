```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
year = today.strftime("%Y")
month = today.strftime("%m")
day = today.strftime("%d")
week_day = today.strftime("%A")
print(f"Today is {week_day}, {month} {day}, {year}")
```

# My birthday

```python
birthday = datetime.date(2002, 7, 19)
birthday_format = birthday.strftime("%A, %B %d, %Y")
print(f"My birthday is {birthday_format}")
```

```python
ninety_d = today + datetime.timedelta(days=90)
ninety_d_month = ninety_d.strftime("%m")
ninety_d_day = ninety_d.strftime("%d")
ninety_d_year = ninety_d.strftime("%Y")

print(f"90 days from today {ninety_d_month} {ninety_d_day}, {ninety_d_year}")
```

# Time for Dinner

```python
dinner_time = datetime.time(17, 30)
hour_dinner_time = dinner_time.strftime("%I, %M, %p")
print(f"Let's met for dinner at {hour_dinner_time}")
```
